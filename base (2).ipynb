### Camada Bronze: dados brutos ingeridos a partir do arquivo history.csv

#### Descrição das variáveis contidas no conjunto de dados

- **CustomerID (ID_Cliente)**: Identificador único de cada cliente

- **Gender (Gênero)**: ndica o gênero do cliente, podendo assumir os valores Masculino (Male) ou Feminino (Female).

- **SeniorCitizen (Idoso)**: Indica se o cliente é ou não uma pessoa idosa. O valor 0 representa que não é idoso e 1 representa que é idoso.

- **Partner (Cônjuge)**: Indica se o cliente possui cônjuge ou parceiro. Pode assumir os valores Sim (Yes) ou Não (No).

- **Dependents (Dependentes)**: Indica se o cliente possui dependentes. Pode assumir os valores Sim (Yes) ou Não (No).

- **Tenure (Tempo_Assinatura)**: Representa o tempo, em meses, que o cliente permanece como assinante do serviço.

- **PhoneService (Servico_Telefonia)**: Indica se o cliente possui serviço de telefonia. Pode assumir os valores Sim (Yes) ou Não (No).

- **MultipleLines (Varias_Linhas)**: Indica se o cliente possui mais de uma linha telefônica. Pode apresentar as opções Sim (Yes), Não (No) ou Sem serviço de telefonia (No phone service).

- **InternetService (Servico_Internet)**: Indica o tipo de serviço de internet utilizado pelo cliente. As categorias presentes na base são DSL, Fibra óptica (Fiber optic) e Sem serviço de internet (No).

- **OnlineSecurity (Seguranca_Online)**: Indica se o cliente possui o serviço de segurança online. Pode apresentar Sim (Yes), Não (No) ou Sem serviço de internet (No internet service).

- **OnlineBackup (Backup_Online)**: Indica se o cliente possui serviço de backup online. Pode apresentar Sim (Yes), Não (No) ou Sem serviço de internet (No internet service).

- **DeviceProtection (Protecao_Dispositivos)**: Indica se o cliente possui proteção para seus dispositivos. Pode apresentar Sim (Yes), Não (No) ou Sem serviço de internet (No internet service).
- **TechSupport (Suporte_Tecnico)**: Indica se o cliente possui serviço de suporte técnico. Pode apresentar Sim (Yes), Não (No) ou Sem serviço de internet (No internet service).
- **StreamingTV (Streaming_TV)**: Indica se o cliente utiliza o serviço de streaming de televisão. Pode apresentar Sim (Yes), Não (No) ou Sem serviço de internet (No internet service).

- **StreamingMovies (Streaming_Filmes)**: Indica se o cliente utiliza o serviço de streaming de filmes. Pode apresentar Sim (Yes), Não (No) ou Sem serviço de internet (No internet service).

- **Contract (Contrato)**: Indica o tipo de contrato firmado pelo cliente. As categorias são Mensal (Month-to-month), Um ano (One year) e Dois anos (Two year).

- **PaperlessBilling (Faturamento_Eletronico)**: Indica se o cliente utiliza faturamento eletrônico, sem o recebimento de fatura impressa. Pode assumir os valores Sim (Yes) ou Não (No).
- **PaymentMethod (Forma_Pagamento)**: Indica a forma utilizada pelo cliente para realizar os pagamentos. As opções incluem Cheque eletrônico (Electronic check), Cheque enviado pelo correio (Mailed check), Transferência bancária (Bank transfer) e Cartão de crédito (Credit card).

- **MonthlyCharges (Valores_Mensais)**: Representa o valor cobrado mensalmente do cliente pelo serviço contratado.

- **TotalCharges (Valores_Totais)**: Representa o valor total cobrado do cliente durante o período em que ele permaneceu como assinante.

- **Churn (Status_Cancelamento)**: Indica se o cliente cancelou ou não o serviço. Yes representa que o cliente cancelou o serviço e No representa que permaneceu como cliente.

- **CustomerFeedbackChurn (Feedback_Cliente)**: Contém um texto com a opinião ou experiência relatada pelo cliente em relação ao serviço, incluindo aspectos como qualidade da internet, preço, forma de pagamento e satisfação geral.

- **MonthlyIncome (Renda_Mensal)**: Representa a renda mensal do cliente.
