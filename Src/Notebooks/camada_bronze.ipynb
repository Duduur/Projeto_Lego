- **Base de dados**

A Terra Signal é uma empresa fictícia do setor
de telecomunicações. A base history.csv,
disponibilizada pela Databricks, contém
informações sobre sua carteira histórica de clientes,
incluindo características individuais, serviços
contratados, relacionamento, informações financeiras
e histórico de cancelamento.

- **Variáveis da Base**

- *Cliente*
    - customerID: ID do cliente;
    - gender: Gênero;
    - SeniorCitizen: É idoso?;
    - Partner: É parceiro?;
    - Dependents: Tem dependentes?;

- *Relacionamento*
    - Tenure: A duração da posse.

- *Serviços*
    - PhoneServico: Serviço Telefonico - Possui?;
    - MultipleLines: Múltiplas linhas - Possui?;
    - InternetService: Serviço de internet;
    - OnlineSecurity: Segurança Online;
    - OnlineBackup: JBackup Online;
    - DeviceProtection: Proteção do dispositivo;
    - TechSupport: Suport técnico;
    - StreamingTV: Streaming na TV;
    - StreamingMovies: Filmes no streaming.

- *Contrato e Pagamento*
    - Contract: Contrato;
    - PaperlessBilling: Faturamento sem papel (Cobrança digital);
    - PaymentMethod: Método de pagamento.

- *Financeiro*
    - MonthlyCharges: Cobranças mensais;
    - TotalCharges: Cobranças totais;
    - MonthlyIncome: Renda mensal.

- *Experiência e Cancelamento*
    - CustomerFeedback: Feedback do cliente;
    - Churn: Taxa de cancelamento.
